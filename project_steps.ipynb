# Steps to Complete the Assignment

## Step 1 — Understand the Dataset
- Load the Metacritic dataset into a Jupyter Notebook.
- Inspect columns, data types, and missing values.
- Review descriptive statistics.
- Identify potential target variables for your business question.

**Deliverables:**
- Dataset overview (`df.info()`, `df.describe()`)
- Notes on potential targets

---

## Step 2 — Formulate the Business Question
- Create one main business question relevant to the Metacritic dataset.
- Define 3–5 related sub-questions.
- Ensure the questions can be answered with machine learning models.

**Deliverables:**
- Written business question
- Written sub-questions

---

## Step 3 — Exploratory Data Analysis (EDA)
- Visualize distributions (histograms, boxplots).
- Compute correlation matrix.
- Identify and remove outliers where necessary.
- Handle missing values.
- Explore text fields, such as review titles or descriptions.

**Deliverables:**
- EDA plots
- Clean dataset

---

## Step 4 — Feature Engineering
- Scale or normalize numerical features.
- Encode categorical variables.
- Create new features if useful.
- Extract text embeddings using a Hugging Face transformer model.

**Deliverables:**
- Feature-engineered dataset
- Code for embedding extraction

---

## Step 5 — Feature Selection
- Apply a matrix decomposition method (e.g., PCA, SVD).
- Reduce dimensionality and select the most meaningful features.

**Deliverables:**
- Reduced feature set
- Explanation of selected features

---

## Step 6 — Dataset Splitting
- Split data into training, validation, and test sets according to best practices.

**Deliverables:**
- Train dataset
- Validation dataset
- Test dataset

---

## Step 7 — Model Fitting
Train and tune the following models using grid search:

1. Baseline model (justify your choice)
2. Random forest model
3. Neural network model

- Track training and validation loss for the neural network.
- Identify the best number of epochs.

**Deliverables:**
- Trained models
- Hyperparameter search results
- Loss-over-epochs plots

---

## Step 8 — Model Evaluation
- Evaluate each model using appropriate performance metrics.
- Create:
  - Actual vs. predicted plots
  - Confusion matrix (if classification)
- Compare how well each model performs on the test set.

**Deliverables:**
- Evaluation metrics
- Diagnostic plots

---

## Step 9 — Model Selection
- Decide which model performs best overall.
- Give justification based on metrics and interpretability.

**Deliverables:**
- Final model choice
- Comparison summary

---

## Step 10 — Explainability Analysis
Apply explainability techniques to the best model:

- SHAP values  
- Counterfactual explanations  

Interpret the results clearly.

**Deliverables:**
- SHAP plots
- Counterfactual examples
- Interpretation summary

---

## Step 11 — Bonus (Optional)
### Option A — Topic Modeling
- Use BERTopic or a similar transformer-based clustering method.
- Identify major topics in review text.
- Evaluate whether the topics correlate with your target variable.

### Option B — Cluster Analysis
- Perform clustering on the dataset.
- Identify whether meaningful groups emerge.

**Deliverables:**
- Topic model or cluster results (optional)

---

## Step 12 — Prepare Deliverables
### Required:
1. **ZIP file** with all notebooks (cleaning, feature engineering, modeling, evaluation).
2. **PowerPoint presentation**.
3. **Video presentation (max 10 minutes)** containing:
   - Problem overview
   - Methods
   - Results
   - Explainability
   - Individual contributions
